```python
from langchang.text_spliter import RecursiveCharacterTextSpliter

from langchang.community.document_loaders import PYPDFLoader, PYPDFDirectoryLoader
from langchang.openai import ChatOpenAI
from langchang.embeddings import OpenAIEmbeddings
from langchang.vectorstores import Chroma

import json
import os

llm = ChatOpenAI(
    model_name="gpt-4o-mini",
    temperature=0.0
    top_p=1.0,
    stop_sequences=['INST']
)


loader = PYPDFDirectoryLoader(path="./data", glob="*.pdf")

txt_splitter = RecursiveCharacterTextSpliter.from_tiktoken_encoder(
    embedding_name="clk1000_base",
    chunk_size=1000,
    chunk_overlap=100
)

research_chunks = loader.load_and_split(txt_splitter)

embeddings =  OpenAIEmbeddings(model_name="text-embedding-3-small")


vector_store = Chroma.from_documents(
    research_chunks,
    embedding=embeddings,
    collection_name="research_papers",
)

reteriever = vector_store.as_retriever(search_kwargs={"k": 3}, search_kwarss={"K": 10})


```
